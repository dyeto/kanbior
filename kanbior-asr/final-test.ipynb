```python
from faster_whisper import WhisperModel
from IPython.display import Audio, display

model = WhisperModel("dyeto/kanbior-fon-asr-v1-ct2", device="cpu", compute_type="int8")  

path = "sample-data/audio/transcribed/tr_000015.wav"
segments, _ = model.transcribe(path, language="fr", beam_size=1, without_timestamps=True)
print(" ".join(s.text.strip() for s in segments))

display(Audio(path, autoplay=False))
```

```python
from transformers import pipeline

asr = pipeline("automatic-speech-recognition", model="dyeto/kanbior-fon-asr-v1")
print(asr(path, generate_kwargs={"language": "fr", "task": "transcribe"})["text"])
```
