```python
from IPython.display import Audio, display
import csv
from pathlib import Path
data_dir = Path("../kanbior-asr/sample-data")
file_path = "audio/transcribed"

with open(data_dir / "transcribed.csv", newline="", encoding="utf-8") as f:
    transcripts = {row["id"]: row for row in csv.DictReader(f)}
```

```python
def get_one(id):
    path = Path(data_dir) / file_path / f"{id}.wav"
    return Audio(str(path), autoplay=False), transcripts[id]["transcription"]
```

```python
a,t = get_one("tr_000002")
display(a)
t
```

```python
lt=""
for i in range(5):
    a, t = get_one(f"tr_00000{i+1}")
    lt = lt + "\n |" +t 

print(lt)
```

```python
from transformers import VitsModel, AutoTokenizer
import torch

model = VitsModel.from_pretrained("facebook/mms-tts-fon")
tokenizer = AutoTokenizer.from_pretrained("facebook/mms-tts-fon")
```

```python
inputs = tokenizer(lt, return_tensors="pt")

with torch.no_grad():
    output = model(**inputs).waveform
    

Audio(output, rate=model.config.sampling_rate)
```
